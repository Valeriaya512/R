## 圖表種類
| 目的 / 資料類型  | 適合圖表     | R 基本函數                  | ggplot2 函數                 | 例子                           |
| ---------- | -------- | ----------------------- | -------------------------- | ---------------------------- |
| 類別比較（定性變數） | 直條圖、橫條圖  | barplot()               | geom_bar()                 | 比較各部門人數。        |
| 次數分布（定性變數） | 長條圖（bar） | barplot()               | geom_bar()                 | 不同性別人數分布。       |
| 數值分布（定量變數） | 直方圖      | hist()                  | geom_histogram()           | 看成績分布。        |
| 數值分布 & 離群值 | 箱型圖（盒形圖） | boxplot()               | geom_boxplot()             | 看不同組別的收入分布。     |
| 時間趨勢       | 折線圖      | plot()（折線）              | geom_line()                | 每月銷售趨勢。              |
| 面積趨勢       | 面積圖      | plot() + polygon()      | geom_area()                | 累積營收變化。                 |
| 兩個連續變數關聯   | 散佈圖（散點圖） | plot()                  | geom_point()               | 身高 vs 體重。 |
| 點圖（單一變數）   | 點圖       | stripchart(), dotplot() | geom_point()               | 簡單數值比較。          |
| 占比         | 圓餅圖      | pie()                   | geom_bar() + coord_polar() | 各產品占總營收比例。            |
| 累計分布       | 累計次數多邊圖  | 自行繪製                    | geom_line()                | 累計次數曲線。         |
| 多指標比較      | 雷達圖      | 需額外套件                   | geom_polygon()             | 產品多面向能力評分。             |
| 地理分布       | 地圖       | 需額外套件（如 sf, tidygraph）  | 需 geom_sf()                | 各縣市銷售比較。              |

## 常見主題
| 主題                 | 特徵             | 適合用途      |
| ------------------ | -------------- | --------- |
| `theme_gray()`     | 預設主題，灰色背景與白色網格 | 一般資料探索    |
| `theme_bw()`       | 白底黑框，網格清晰      | 論文、報告     |
| `theme_minimal()`  | 白底、簡潔、淡網格      | 簡報、現代風格   |
| `theme_classic()`  | 僅保留座標軸，無網格     | 學術期刊常見    |
| `theme_light()`    | 淺灰網格，較柔和       | 商業報表      |
| `theme_dark()`     | 深色背景           | 投影片、深色介面  |
| `theme_void()`     | 幾乎移除所有元素       | 製作地圖、資訊圖表 |
| `theme_linedraw()` | 強調線條與邊框        | 印刷用途      |
| `theme_test()`     | 測試用主題          | 開發與除錯     |

## 長條圖(geom_col)

```r
# 統計各汽缸數量
library(ggplot2)
library(dplyr)

cyl_data <- mtcars %>%
  count(cyl)

ggplot(cyl_data,
       aes(x = factor(cyl),
           y = n)) +
  geom_col(
    fill = "steelblue",
    color = "black",
    width = 0.7
  ) +
  labs(
    title = "汽缸數量分布",
    x = "Cylinders",
    y = "Count"
  ) +
  theme_minimal()
```

## 折線圖(geom_line)

```r
ap <- data.frame(
  Year = as.numeric(time(AirPassengers)),
  Passengers = as.numeric(AirPassengers)
)

ggplot(ap,
       aes(Year, Passengers)) +
  geom_line(
    color = "tomato",
    linewidth = 1.2
  ) +
  labs(
    title = "Air Passengers Trend"
  ) +
  theme_minimal()
```

## 折線圖+點

```r
ap <- data.frame(
  Year = as.numeric(time(AirPassengers)),
  Passengers = as.numeric(AirPassengers)
)

ggplot(ap,
    aes(Year, Passengers)) +
  geom_line(
    color = "#2C7FB8",
    linewidth = 1.2
  ) +
  geom_point(
    color = "#D95F0E",
    size = 2
  ) +
  theme_minimal()
```

## 散點圖(geom_point)

```r
#看車重與油耗關係
library(ggplot2)

ggplot(mtcars,
    aes(wt, mpg)) +
  geom_point(
    color = "darkgreen",
    size = 3
  ) +
  labs(
    title = "Weight vs MPG"
  ) +
  theme_minimal()
```

## 散點圖+趨勢線

```r
library(ggplot2)

ggplot(data = mtcars, aes(x = wt, y = mpg)) +
  geom_point(color = "steelblue", size = 2) +
  geom_smooth(method = "lm", color = "red", se = FALSE) +  # 趨勢線
  labs(
    title = "Scatter plot: wt vs mpg",
    x = "Weight (wt)",
    y = "MPG (mpg)"
  ) +
  theme_minimal()
```

```r
ggplot(mtcars,
    aes(wt, mpg)) +
  geom_point(
    color = "navy",
    size = 3
  ) +
  geom_smooth(
    method = "lm",
    color = "red",
    linewidth = 1.2,
    se = FALSE
  ) +
  theme_minimal()
```

## 直方圖(histogram)

```r
library(ggplot2)

ggplot(data = mtcars, aes(x = mpg)) +
  geom_histogram(binwidth = 5, fill = "skyblue", color = "black") +
  labs(
    title = "Histogram: MPG distribution",
    x = "MPG",
    y = "Count"
  ) +
  theme_minimal()
```

## 箱型圖(boxplot)

```r
library(ggplot2)

# 將 cyl 轉換為 factor 類別變數以正確分組
ggplot(data = mtcars, aes(x = factor(cyl), y = mpg, fill = factor(cyl))) +
  geom_boxplot(color = "black") +
  labs(
    title = "Boxplot: MPG by Cylinder",
    x = "Number of Cylinders",
    y = "MPG",
    fill = "Cylinders"
  ) +
  theme_minimal()
```

```r
# 自訂顏色
ggplot(iris,
    aes(Species,
      Sepal.Length,
      fill = Species)) +
  geom_boxplot() +
  scale_fill_manual(
    values = c(
      "setosa" = "#1B9E77",
      "versicolor" = "#D95F02",
      "virginica" = "#7570B3"
    )
  ) +
  theme_minimal()
```

## 密度圖(density plot)

```r
ggplot(iris,
    aes(
      x = Sepal.Length,
      fill = Species
      )) +
  geom_density(
    alpha = 0.4  # 透明度
  ) +
  theme_minimal()
```

## 分組長條圖

```r
iris_count <- iris %>%
  count(Species)

ggplot(iris_count,
    aes(
      Species,
      n,
      fill = Species
      )) +
  geom_col(
    color = "black"
  ) +
  theme_minimal()
```

## 常見報表風格

```r
ggplot(mtcars,
    aes(
      x = reorder(rownames(mtcars), mpg),
      y = mpg
      )) +
  geom_col(
    fill = "#2C7FB8"
  ) +
  coord_flip() +
  labs(
    title = "MPG by Car Model",
    x = "",
    y = "MPG"
  ) +
  theme_minimal(base_size = 14)
```
