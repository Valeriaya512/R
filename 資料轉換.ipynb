## dplyr套件

| 函式              | 轉成什麼       | 常見用途            |
| --------------- | ---------- | --------------- |
| as_tibble()     | 一般 tibble  | 把現有資料框、矩陣、列表轉格式 |
| as_tibble_row() | 1 列 tibble | 把一個向量變成「一筆資料」   |

```r
library(tibble)

df <- data.frame(a = 1:3, b = c("x", "y", "z"))
as_tibble(df)  #把data frame、matrix、list等物件轉成tibble，常用在你已經有一個資料物件，想改用tidyverse風格處理

as_tibble_row(c(a = 1, b = 2, c = 3))  #把一個向量變成單列tibble，每個元素會變成一個欄位

as_tibble_col(1:3)  #把向量變成單欄tibble，適合想把一串值包成一個欄位時使用

as_tibble_col(1:3, column_name = "score")
```

## mutate()
新增或修改欄位，但保留原本所有欄位

功用：
*  新增欄位
*  修改欄位
*  刪除欄位

常見用法：
* 建立計算欄位：mutate(df, total = a + b)。
* 修改原欄位：mutate(df, a = a * 2)。

* 搭配條件判斷：if_else()、case_when()。

* 搭配位移函式：lag()、lead()。

* 做累積計算：cumsum()、cummean()。

```r
# 新增欄位
library(dplyr)
library(tibble)

iris.tb <- as_tibble(iris)

iris.tb <- iris_tb %>%
  mutate(
  Sepal_Ratio = Sepal.Length / Sepal.Width ,
  Sepal_Ratio_cm = Sepal_Ratio * 2.54
  )
names(iris.tb)
print(head(iris.tb,10))
```

```r
# 用if_else()建立分類欄位
library(dplyr)
library(tibble)

iris.tb <- as_tibble(iris)

iris.tb <- iris.tb %>%
  mutate(size = if_else(Sepal.Length > 5, "big", "small"))

print(head(iris.tb,10))
```

```r
# 用case_when()做多條件分類
library(dplyr)
library(tibble)

iris.tb <- as_tibble(iris)

iris.tb <- iris.tb %>%
  mutate(size = case_when(
    Sepal.Length < 5 ~ "small",
    Sepal.Length < 6 ~ "medium",
    TRUE ~ "large"
  ))

print(head(iris.tb,10))
```

```r
library(dplyr)
library(tibble)

iris.tb <- as_tibble(iris)

# 取對數或其他轉換
iris.tb1 <- iris.tb %>%
  mutate(log_len = log(Sepal.Length))

print(head(iris.tb1))


# 做比例或標準化
iris.tb2 <- iris.tb %>%
  mutate(prop = Sepal.Length / sum(Sepal.Length))

print(head(iris.tb2))
```

```r
library(dplyr)
library(tibble)

iris.tb <- as_tibble(iris)
#iris.tb

# lag()：看前一筆
iris.tb1 <- iris.tb %>%
  arrange(Sepal.Length) %>%
  mutate(prev_len = lag(Sepal.Length, n = 2))  # n=2 往前看2筆
#  mutate(prev_len = lag(Sepal.Length, default = 0))  # default = 0：NA補0

print(head(iris.tb1))

# lead()：看下一筆
iris.tb2 <- iris.tb %>%
  arrange(Sepal.Length) %>%
  mutate(next_len = lead(Sepal.Length, n = 2))  # 往後看2筆

print(tail(iris.tb2))
```

```r
# 修改欄位
library(dplyr)
library(tibble)

iris.tb <- as_tibble(iris)

iris.tb <- iris.tb %>%
  mutate(
  Sepal.Length = Sepal.Length * 10
  )
print(head(iris.tb,10))
```

```r
# 刪除欄位
library(dplyr)
library(tibble)

iris.tb <- as_tibble(iris)

iris.tb <- iris.tb %>%
  mutate(
  Sepal.Length = NULL
  )

print(head(iris.tb,10))
```

| 函式          | 主要作用    | 會保留原欄位嗎       | 常見用途      |
| ----------- | ------- | ------------- | --------- |
| select()    | 挑選欄位    | 會，只是把沒選到的欄位移除 | 只看需要的變數   |
| mutate()    | 新增/修改欄位 | 會             | 做計算、分類、轉換 |
| transmute() | 新增/修改欄位 | 不會            | 只輸出計算後結果  |

```r
# transmute 函數：新增或修改欄位後，只留下新欄位，類似「mutate() + select()」的組合
library(dplyr)
library(tibble)

iris.tb <- as_tibble(iris)

iris.tb1 <- iris.tb %>%
  transmute(ratio = Sepal.Length / Sepal.Width)

iris.tb1
```

## group_by()
用來分組\
*常和 summarise()、mutate() 搭配使用*

```r
library(dplyr)
library(tibble)

iris.tb <- as_tibble(iris)

iris.tb <- iris.tb %>%
  group_by(Species) %>%
  do(browser())

iris.tb
```

```r
library(dplyr)
library(tibble)

trees.tb <- as_tibble(trees)

trees.tb1 <- trees.tb %>%
  group_by(Girth) %>%
  summarise(
  number = n(),
  avg_Height = mean(Height),
  sd_Height = sd(Height),
  max_Height = max(Height),
  min_Height = min(Height)
  ) %>%
    arrange(desc(avg_Height))

trees.tb1
```

```r
library(dplyr)
library(tibble)

iris.tb <- as_tibble(iris)

iris.tb1 <- iris.tb %>%
  group_by(Species) %>%
  filter(rank(desc(Sepal.Length)) < 4) %>%  #各組前3名
  arrange(desc(Species),desc(Sepal.Length))

iris.tb1
```

##　character vs factor vs ordered factor
* character：只把資料當文字，R 不會自動把它理解成分類，例如姓名、地址、留言
* factor：存放「分類資訊」，會把不同值整理成**有限個** levels，例如性別、血型、教育程度、地區等，讓 R 知道這些值是不同的類別，方便統計分析、建模和繪圖
* ordered factor：除了分類，還明確表示順序，例如 low < medium < high

| 類型             | 用途          | 有沒有順序 | 典型例子      | 常見用途       |
| -------------- | ----------- | ----- | --------- | ---------- |
| character      | 存文字內容本身     | 沒有    | 姓名、地址、備註  | 文字處理、拼接、搜尋 |
| factor         | 存「類別」資料     | 沒有    | 性別、血型、城市  | 分組、統計次數、建模 |
| ordered factor | 存「有高低順序」的類別 | 有     | 等級、滿意度、成績 | 排序、比較、序位分析 |

```r
# character
name <- c("Amy", "Ben", "Cindy")
name
```

```r
# factor
gender <- c("male", "female", "female", "male")
gender_f <- factor(gender)
gender_f
```

```r
# ordered factor
level <- c("B", "A", "C", "A")
level_f <- factor(level, levels = c("C", "B", "A"), ordered = TRUE)
level_f
```
