## dplyr套件

| 函式              | 轉成什麼       | 常見用途            |
| --------------- | ---------- | --------------- |
| as_tibble()     | 一般 tibble  | 把現有資料框、矩陣、列表轉格式 |
| as_tibble_row() | 1 列 tibble | 把一個向量變成「一筆資料」   |
| as_tibble_col() | 1 欄 tibble | 把一個向量變成「一個欄位」   |

```r
library(tibble)

df <- data.frame(a = 1:3, b = c("x", "y", "z"))
as_tibble(df)  #把data frame、matrix、list等物件轉成tibble，常用在你已經有一個資料物件，想改用tidyverse風格處理

as_tibble_row(c(a = 1, b = 2, c = 3))  #把一個向量變成單列tibble，每個元素會變成一個欄位

as_tibble_col(1:3)  #把向量變成單欄tibble，適合想把一串值包成一個欄位時使用

as_tibble_col(1:3, column_name = "score")
```

## 資料篩選與排序
*  filter()：用來篩選列
*  select()：用來篩選欄
*  arrange()：排序

```r
library(dplyr)
library(tibble)

iris.tb <- as_tibble(iris) #as_tibble()：把資料整理成tidyverse常用的表格格式
print(iris.tb)
```

```r
library(dplyr)

#f <- filter(iris.tb, Sepal.Length > 7, Petal.Length > 6)
f <- iris.tb %>% filter(Sepal.Length > 7, Petal.Length > 6)
print(f)
```

```r
s <- select(iris.tb, Sepal.Length, Sepal.Width)
print(head(s))
```

```r
iris.tb <- iris.tb %>%
#  filter(Sepal.Length > 5.5, Sepal.Width > 3.3, Species == 'setosa') %>%
  filter(Sepal.Length > 5.5, Sepal.Width > 3.3) %>%
  select(Species)

iris.tb
```

```r
iris.tb <- iris.tb %>%
  arrange(Sepal.Length, Sepal.Width) #預設升冪

print(head(iris.tb,15))
```

```r
iris.tb <- iris.tb %>%
  arrange(desc(Sepal.Length), Sepal.Width)  #降冪

print(head(iris.tb,15))
```

## mutate()
功用：
*  新增欄位
*  修改欄位
*  刪除欄位

常見用法：
* 建立計算欄位：mutate(df, total = a + b)。
* 修改原欄位：mutate(df, a = a * 2)。

* 搭配條件判斷：if_else()、case_when()。

* 搭配位移函式：lag()、lead()。

* 做累積計算：cumsum()、cummean()。

```r
# 新增欄位
library(tibble)

iris.tb <- as_tibble(iris)

iris.tb <- iris_tb %>%
  mutate(
  Sepal_Ratio = Sepal.Length / Sepal.Width ,
  Sepal_Ratio_cm = Sepal_Ratio * 2.54
  )
names(iris.tb)
print(head(iris.tb,10))
```

```r
# 用if_else()建立分類欄位
library(tibble)

iris.tb <- as_tibble(iris)

iris.tb <- iris.tb %>%
  mutate(size = if_else(Sepal.Length > 5, "big", "small"))

print(head(iris.tb,10))
```

```r
# 修改欄位
library(tibble)

iris.tb <- as_tibble(iris)

iris.tb <- iris.tb %>%
  mutate(
  Sepal.Length = Sepal.Length * 10
  )
print(head(iris.tb,10))
```

```r
# 刪除欄位
library(tibble)

iris.tb <- as_tibble(iris)

iris.tb <- iris.tb %>%
  mutate(
  Sepal.Length = NULL
  )

print(head(iris.tb,10))
```
