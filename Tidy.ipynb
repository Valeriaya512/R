```r
# tibble()：用欄(column)來建立資料表
tb1 <- tibble(
  id    = 1:5,
  name  = c("Alice", "Bob", "Charlie", "David", "Eve"),
  score = c(85, 90, 78, 92, 88)
)
tb1

tb2 <- tibble(x = 1:5, y = 1, z = x ^ 2 + y)
tb2
```

```r
# tibble傳出表格
tb1[[2]]  #加[]向量
```

```r
# tribble()：用列(row)來輸入少量資料
trb3 <- tribble(
  ~x, ~y, ~z,  #欄位名稱前面要加"~"
  "a", 2, 3.6,
  "b", 1, 8.9
)
trb3
```

```r
# 增加列
df <- tibble(x = 1 : 3, y = 3 : 1)
add_row(df, x = 4, y = 0)
```

```r
# 在第二列插入新的列
df <- tibble(x = 1 : 3, y = 3 : 1)
add_row(df, x = 4, y = 0, .before = 2) #.before <- 要插入到第幾列
```

```r
# 插入多個列
df <- tibble(x = 1 : 3, y = 3 : 1)
add_row(df, x = 4 : 6, y = 0 : -2)
```

```r
# 插入的列可以有缺失值
df <- tibble(x = 1 : 3, y = 3 : 1)
#add_row(df, x = 4)
add_row(df, y = 0)
```

```r
# 增加欄
df <- tibble(x = 1 : 3, y = 3 : 1)
add_column(df, z = -1 : 1, w = 0)
```

## Tibble 與 data.frame 的差異
| 項目             | data.frame                             | tibble                                                     |
| -------------- | -------------------------------------- | ---------------------------------------------------------- |
| 列印顯示           | 會顯示所有行與列， datum 多時控制台會很長、很亂            | 只顯示前 10 行和還能納進控制台的欄位，輸出較整潔易讀 sohu+1                        |
| 欄位名稱部分匹配       | df$col 或 df[, "co"] 可能部分匹配，不嚴格         | 部分匹配會直接報錯：Unknown or uninitialised column listendata       |
| 單欄提取結果         | df[, "col"] 會回傳向量（不是 data.frame）       | tb[, "col"] 仍然回傳 tibble（保持表格結構） listendata                 |
| $ 提取           | df$col 回傳向量                            | tb$col 也回傳向量，但 tb[["col"]] 更嚴格                             |
| 新增欄位長度檢查       | 可以新增長度不同的欄位，R 會自動重複（回收）                | 新增欄位長度必須與現有行數一致，否則直接報錯 listendata                          |
| ** factor 轉換** | 預設可能把字串自動轉成 factor（舊版R）                | 不會自動把字串轉成 factor，字串保持為 character csdn                      |
| 建構語法           | data.frame(col1 = ..., col2 = ...)     | tibble(col1 = ..., col2 = ...) 更簡潔，支援自動回收長度 1 向量 csdn      |
| 與 tidyverse 整合 | 多數 tidyverse 函數也能用，但有時會自動轉回 data.frame | 與 dplyr、tidyr、readr 等緊密整合，輸出通常保持 tibble sohu               |
| 性能             | 處理大資料時較慢，尤其在複雜操作時                      | 底層結構更優化，但實際差異不大；大部分速度差異來自 tidyverse 函數，而非 tibble 本身 sohu+2 |
| 支援複雜結構         | 不建議也不擅長儲存 list-columns                 | 支援 list-columns，可儲存複雜結構（如向量、列表） listendata                 |

```r
# data.frame：50 行會全部顯示（或至少很多）
df <- data.frame(
  x = 1:50,
  y = 1:50 * 2
)
df

# tibble：預設只顯示前 10 行
tb <- tibble(
  x = 1:50,
  y = 1:50 * 2
)
tb
```

```r
# 單欄提取結果
df <- data.frame(
  x = 1:5,
  y = 6:10
)

tb <- tibble(
  x = 1:5,
  y = 6:10
)

# data.frame：取單欄會變成向量
df[, "x"]     # 是 vector
class(df[, "x"])

# tibble：取單欄仍然是 tibble
tb[, "x"]     # 是 tibble
class(tb[, "x"])
```

```r
# 新增欄為長度檢查
# data.frame 會自動回收，但 tibble 嚴格檢查，避免隱含錯誤


# data.frame：可以新增不同長度，R 會自動重複
df <- data.frame(x = 1:5)
df$newcol <- c(5, 6)  # 長度 2 vs 5 行，會重複
df

# tibble：長度不一致會直接報錯
tb <- tibble(x = 1:5)
tb$newcol <- c(5, 6)  # 錯誤：長度必須匹配
```

```r
# factor轉換

# data.frame（舊版 R）：預設可能把字串轉成 factor
df <- data.frame(
  name = c("Alice", "Bob")
)
class(df$name)  # 可能是 factor

# tibble：字串保持為 character
tb <- tibble(
  name = c("Alice", "Bob")
)
class(tb$name)  # character
```

```r
# 將資料從tibble轉換成data.frame
tb <- tibble(
  x = 1:50,
  y = 1:50 * 2
)

class(as.data.frame(tb))
```
